```python
import pandas as pd
from IPython.display import display, HTML

# Define the data
data1 = {
    ('Target Name', ''): ["Sand", "Silt", "Clay", "TOC"],
    ('Method', ''): ["Cubist", "Cubist", "Kernel Ridge Regression", "SVR"],
    ('Model Details', 'Model Name'): ["sand_model_cubist", "silt_model_cubist", "clay_model_ridge", "TOC_model_svr"],
    ('Model Details', 'Spectral Preprocessing'): ["none", "fod", "fod_log", "log"],
    ('R2 Scores', 'Calibration'): ["0.39", "0.26", "0.31", "0.43"],
    ('R2 Scores', 'Validation'): ["0.24", "0.22", "0.21", "0.40"],
    ('R2 Scores', 'Leave one out'): ["0.31", "0.20", "0.26", "0.42"],
}

# Create the DataFrame using MultiIndex for columns
df1 = pd.DataFrame(data1)

# Style the DataFrame for display in Jupyter Notebook
styled_df1 = df1.style.set_table_styles({
    ('Model Details', 'Model Name'): [{'selector': 'th', 'props': [('text-align', 'center')]}],
    ('Model Details', 'Spectral Preprocessing'): [{'selector': 'th', 'props': [('text-align', 'center')]}],
    ('R2 Scores', 'Calibration'): [{'selector': 'th', 'props': [('text-align', 'center')]}],
    ('R2 Scores', 'Validation'): [{'selector': 'th', 'props': [('text-align', 'center')]}],
    ('R2 Scores', 'Leave one out'): [{'selector': 'th', 'props': [('text-align', 'center')]}],
}).set_properties(**{'text-align': 'center'})

# Display the heading and DataFrame
display(HTML("<h2>UAE Venus</h2>"))
display(styled_df1)
```

```python
# Convert DataFrames to HTML with headings
html_content = f"""
<html>
<head><title>Data Export</title></head>
<body>
    <h2 style='text-align: center;'>UAE Venus</h2>
    {styled_df1.to_html(index=False)}
</body>
</html>
"""

# Write to an HTML file
with open('UAE_Venus.html', 'w') as f:
    f.write(html_content)

print("HTML file with headings generated successfully.")
```
