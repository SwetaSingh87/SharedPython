```python
import os
import shutil

# Step 1: Create the 'pkl' folder in the current directory
current_directory = os.getcwd()  # Get the current working directory
pkl_folder = os.path.join(current_directory, 'pkl')
os.makedirs(pkl_folder, exist_ok=True)  # Create pkl folder if it doesn't exist

# Define the specific filenames to look for
specific_files = ['data.pickle', 'meta_data.pickle', 'HMtree.pickle']

# Step 2: Loop through folders in the current directory and create corresponding folders in pkl
for item in os.listdir(current_directory):
    item_path = os.path.join(current_directory, item)
    # Check if it's a directory and if it's not the 'pkl' folder
    if os.path.isdir(item_path) and item != 'pkl':
        # Create a folder inside 'pkl' with the same name
        os.makedirs(os.path.join(pkl_folder, item), exist_ok=True)

        # Step 3: Copy specified .pickle files from the current folder to the corresponding pkl folder
        for file in specific_files:  # Loop through the specified file names
            source_file = os.path.join(item_path, file)
            # Check if the specified .pickle file exists in the current directory
            if os.path.isfile(source_file):  
                destination_file = os.path.join(pkl_folder, item, file)
                shutil.copy(source_file, destination_file)  # Copy the file

print("Specified .pickle files have been copied to their respective folders in the 'pkl' directory.")
```
